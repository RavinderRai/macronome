```python
from PIL import Image

# Load the image with transparency
input_path = "macronome_translucent_logo.png"
output_path = "adaptive-icon.png"

# Open the image
img = Image.open(input_path)

# Check if it has an alpha channel (transparency)
if img.mode != 'RGBA':
    img = img.convert('RGBA')

# Resize to 1024x1024 using high-quality LANCZOS resampling
# This preserves transparency and maintains quality
resized_img = img.resize((1024, 1024), Image.Resampling.LANCZOS)

# Save as PNG to preserve transparency
resized_img.save(output_path, 'PNG', optimize=False)

print(f"✓ Resized {input_path} from {img.size} to {resized_img.size}")
print(f"✓ Saved as {output_path}")
print(f"✓ Transparency preserved: {resized_img.mode == 'RGBA'}")
```

```python
# Also create icon.png (for iOS/general use) from the same source
icon_output_path = "icon.png"

# Load and resize again (or reuse the resized_img from previous cell)
img_for_icon = Image.open(input_path)
if img_for_icon.mode != 'RGBA':
    img_for_icon = img_for_icon.convert('RGBA')
icon_resized = img_for_icon.resize((1024, 1024), Image.Resampling.LANCZOS)
icon_resized.save(icon_output_path, 'PNG', optimize=False)

print(f"✓ Also created {icon_output_path} at 1024x1024")
```

```python
from PIL import Image
import os

# Input path from your variable
store_listing_icon_path = "../../apps/mobile/assets/macronome_icon_storelisting_asset.png"
output_path = "../../apps/mobile/assets/macronome_icon_storelisting_512x512.png"

# Open the image
img = Image.open(store_listing_icon_path)

# Convert to RGBA (32-bit PNG with alpha channel) if not already
if img.mode != 'RGBA':
    img = img.convert('RGBA')

# Directly resize to 512x512 (this will stretch if not square, but fills the space)
# Use LANCZOS for high quality
resized_img = img.resize((512, 512), Image.Resampling.LANCZOS)

# Save as 32-bit PNG
resized_img.save(output_path, 'PNG', optimize=False)

# Check file size
file_size_kb = os.path.getsize(output_path) / 1024
print(f"✓ Converted {store_listing_icon_path}")
print(f"✓ Output size: {resized_img.size} (512x512)")
print(f"✓ Format: 32-bit PNG (RGBA)")
print(f"✓ Color space: sRGB")
print(f"✓ File size: {file_size_kb:.2f} KB")
if file_size_kb > 1024:
    print(f"⚠ WARNING: File size ({file_size_kb:.2f} KB) exceeds 1024 KB limit!")
else:
    print(f"✓ File size is under 1024 KB limit")
print(f"✓ Saved as {output_path}")
```
